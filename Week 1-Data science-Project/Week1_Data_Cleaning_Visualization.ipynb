# Week1_Data_Cleaning_Visualization
Data Cleaning and Basic Visualization

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=pd.read_csv('sales_data.csv')
df.head()
```

## Dataset Info

```python
df.info()
df.describe()
```

## Missing Values and Duplicates

```python
print(df.isnull().sum())
df=df.drop_duplicates()
for c in df.columns:
    if 'date' in c.lower():
        df[c]=pd.to_datetime(df[c],errors='coerce')
```

## Visualizations

```python
num=df.select_dtypes(include='number').columns
cat=df.select_dtypes(exclude='number').columns
if len(cat):
    plt.figure(figsize=(6,4));df[cat[0]].value_counts().plot(kind='bar');plt.title('Bar Chart');plt.xlabel(cat[0]);plt.ylabel('Count');plt.show()
if len(num):
    plt.figure(figsize=(6,4));df[num[0]].hist();plt.title('Histogram');plt.xlabel(num[0]);plt.ylabel('Frequency');plt.show()
if len(num)>1:
    plt.figure(figsize=(6,4));df[num[:2]].plot(x=num[0],y=num[1]);plt.title('Line Chart');plt.show()
if len(cat):
    plt.figure(figsize=(5,5));df[cat[0]].value_counts().head().plot(kind='pie',autopct='%1.1f%%');plt.ylabel('');plt.title('Pie Chart');plt.show()
```
